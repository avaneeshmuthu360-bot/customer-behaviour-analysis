```python
customer_data["Review Rating"]=customer_data.groupby("Category")["Review Rating"].transform(lambda x: x.fillna(x.median()))
```

```python
customer_data.columns=customer_data.columns.str.lower()
customer_data.columns=customer_data.columns.str.replace(' ','_')
customer_data=customer_data.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
customer_data.columns
```

# Age Grouping

```python
labels=['Young_adult','adult','Middle_aged','senior']
customer_data["age_category"]=pd.qcut(customer_data["age"],q=4,labels=labels)
```

```python
customer_data[["age","age_category"]].head(10)
```

# Purchase Frequency

```python
frequency_map={
    'fortnightly':14,
    'weekly':7,
    'monthly':30,
    'quarterly':90,
    'bi-weekly':14,
    'annually':365,
    'every 3 months':90
}
customer_data["purchase_frequency_days"]=customer_data["frequency_of_purchases"].str.strip().str.lower().map(frequency_map)
```

```python
customer_data[["frequency_of_purchases","purchase_frequency_days"]].head()
```

```python
customer_data[['discount_applied', 'promo_code_used']].head()
```

```python
(customer_data['discount_applied']==customer_data['promo_code_used']).all()
```

```python
customer_data=customer_data.drop("promo_code_used",axis=1)
```

```python
import os
from sqlalchemy import create_engine

engine = create_engine(r"sqlite:///expense.db")

customer_data.to_sql("customer_behaviour", engine, if_exists="replace", index=False)
engine.dispose()

print(f"Success! Saved to: {os.path.abspath('expense.db')}")
```
